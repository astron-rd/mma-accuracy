```python
import pandas as pd
import matplotlib.pyplot as plt

# Read the data from the file, treating 'N/A' as NaN
df = pd.read_csv('values.txt', sep="\\s+", na_values='N/A')

plt.figure(figsize=(10,6))
for col in df.columns[1:]:
    plt.scatter(df['i'], df[col], label=col)

plt.xlabel('i')
plt.ylabel('Value')
plt.title('Scatter plot of series')
plt.ylim(-1000, 1000)
plt.legend()
plt.grid(True)
plt.show()
```

```python
total_width = 20
ncols = 5
height = 4
aspect = (total_width / ncols) / height

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Read the data
df = pd.read_csv('values.txt', sep=r'\s+', na_values='N/A')

# Melt the dataframe to long format
df_long = df.melt(id_vars='i', var_name='datatype', value_name='value')

# Create a FacetGrid with one facet per 'prec'
g = sns.FacetGrid(df_long, col='datatype', col_wrap=3, sharex=False, sharey=False, height=height, aspect=aspect)
g.map_dataframe(sns.scatterplot, x='i', y='value')


for i, ax in enumerate(g.axes.flatten()):
    if i == 0 or i == 3:
        ax.set_ylabel("Value")
    if i > 1:
        ax.set_xlabel("index")

    ax.set_xscale('log')
    ax.set_yscale('log')
    ax.grid(True, which="both", ls="--", alpha=0.8)

plt.show()
```
